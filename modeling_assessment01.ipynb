## Preprocessing
<strong> Task : Use TF-IDF Vectorization to create a vectorized document term matrix. You may want to explore the max_df and min_df parameters.</strong>

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfidf = TfidfVectorizer(max_df=0.95,min_df=2,stop_words='english')
```

```python
dtm = tfidf.fit_transform(npr['Question'])
```

```python
dtm
```

## NoN-negative Matrix Factorization 
<strong> Task : Using Scikit-learn create an instance of nmf with 20 expected componets.(user random_state=42)

```python
from sklearn.decomposition import NMF
```

```python
nmf_model = NMF(n_components=20,random_state=42)
```

```python
nmf_model.fit(dtm)
```

<strong> Task: print our the top 15 most common word for each of the 20 topics.</strong>

```python
for index,topic in enumerate(nmf_model.components_):
    print(f"The top 15 words for topic # {index}")
    print([tfidf.get_feature_names_out()[i] for i in topic.argsort()[-15:]])
    print('\n')
```

<strong>Task: Add a new column to the original quora dataframe that labels each question into one of the 20 topic categories</strong>

```python
topic_results = nmf_model.transform(dtm)
```

```python
npr['Topic'] = topic_results.argmax(axis=1)
```
